```python
import random
import copy
import math

winMatrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0],
]

data = list(map(int, input("Nhập ma trận: ").split()))
matrix = [data[i:i+3] for i in range(0, 9, 3)]

T0 = float(input("Nhập nhiệt độ ban đầu T0: "))
Tmin = float(input("Nhập nhiệt độ tối thiểu Tmin: "))
alpha = float(input("Nhập hệ số giảm nhiệt alpha: "))
T = T0

father = None
action = None
step = 0

def costCal(thisMatrix):
    hn = 0
    x1 = x2 = y1 = y2 = 0
    for a in range(1,9):
        for i in range(3):
            for j in range(3):
                if thisMatrix[i][j] == a:
                    x1, y1 = i, j
        for i in range(3):
            for j in range(3):
                if winMatrix[i][j] == a:
                    x2, y2 = i, j
        hn = hn + abs(x1 - x2) + abs(y1 - y2)
    return hn

cost = costCal(matrix)

info = [[matrix, father, action, step, cost]]
randomNeighbor = []

done = False

while(T > Tmin):
    if matrix == winMatrix:
        done = True
        print("KẾT QUẢ ")
        print("Start:")
        for row in info[0][0]:
            print(row)

        print("\n")

        print("Goal")
        for row in winMatrix:
            print(row)

        print("\n")

        print("CÁC BƯỚC DI CHUYỂN")
        for i in range(0, len(info)):
            currentState = info[i]
            currMatrix = currentState[0]
            currAct = currentState[2]
            currStep = currentState[3]
            currCost = currentState[4]

            print(f"Step {currStep}:   Action = {currAct}   Cost: {currCost}")
            for row in currMatrix:
                print(row)
            print("\n")
        break

    if info[-1][0] == winMatrix:
        done = True
        print("KẾT QUẢ ")
        print("Start:")
        for row in info[0][0]:
            print(row)

        print("\n")

        print("Goal")
        for row in winMatrix:
            print(row)

        print("\n")

        print("CÁC BƯỚC DI CHUYỂN")
        for i in range(0, len(info)):
            currentState = info[i]
            currMatrix = currentState[0]
            currAct = currentState[2]
            currStep = currentState[3]
            currCost = currentState[4]

            print(f"Step {currStep}:   Action = {currAct}   Cost: {currCost}")
            for row in currMatrix:
                print(row)
        break

    for i in range(3):
        for j in range(3):
            if matrix[i][j] == 0:
                x, y = i, j


    if x > 0:
        newMatrix = copy.deepcopy(matrix)
        newMatrix[x][y] = newMatrix[x-1][y]
        newMatrix[x-1][y] = 0
        childCost = costCal(newMatrix)
        randomNeighbor.append([newMatrix, matrix, "UP", step + 1, childCost])


    if x < 2:
        newMatrix = copy.deepcopy(matrix)
        newMatrix[x][y] = newMatrix[x+1][y]
        newMatrix[x+1][y] = 0
        childCost = costCal(newMatrix)
        randomNeighbor.append([newMatrix, matrix, "DOWN", step + 1, childCost])


    if y > 0:
        newMatrix = copy.deepcopy(matrix)
        newMatrix[x][y] = newMatrix[x][y-1]
        newMatrix[x][y-1] = 0
        childCost = costCal(newMatrix)
        randomNeighbor.append([newMatrix, matrix, "LEFT", step + 1, childCost])


    if y < 2:
        newMatrix = copy.deepcopy(matrix)
        newMatrix[x][y] = newMatrix[x][y+1]
        newMatrix[x][y+1] = 0
        childCost = costCal(newMatrix)
        randomNeighbor.append([newMatrix, matrix, "RIGHT", step + 1, childCost])

    nextState = random.choice(randomNeighbor)
    randomNeighbor = []
    denta = nextState[4] - cost
    if denta < 0:
        matrix = nextState[0]
        step = nextState[3]
        cost = nextState[4]
        info.append(nextState)
    else:
        p = math.exp(-denta / T)
        if random.random() < p:
            matrix = nextState[0]
            step = nextState[3]
            cost = nextState[4]
            info.append(nextState)

    if (T * alpha) <= Tmin:
        print("Start:")
        for row in info[0][0]:
            print(row)
        print("\n")
        print("Goal")
        for row in winMatrix:
            print(row)
        print("\n")
        print("CÁC BƯỚC DI CHUYỂN")
        for i in range(0, len(info)):
            currentState = info[i]
            currMatrix = currentState[0]
            currAct = currentState[2]
            currStep = currentState[3]
            currCost = currentState[4]
            print(f"Step {currStep}:   Action = {currAct}   Cost: {currCost}")
            for row in currMatrix:
                print(row)
            print("\n")

    T = alpha * T
```

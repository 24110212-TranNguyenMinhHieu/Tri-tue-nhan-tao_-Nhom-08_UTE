```python
import copy

winMatrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0],
]

data1 = list(map(int, input("Nhập ma trận thứ nhất: ").split()))
matrix1 = [data1[i:i+3] for i in range(0, 9, 3)]

data2 = list(map(int, input("Nhập ma trận thứ hai: ").split()))
matrix2 = [data2[i:i+3] for i in range(0, 9, 3)]

father = None
action = None
step = 0

def costCal(thisMatrix):
    matrixA, matrixB = thisMatrix
    hn1 = hn2 = hn = 0
    x1 = x2 = y1 = y2 = xg = yg = 0
    for a in range(1, 9):
        for i in range(3):
            for j in range(3):
                if matrixA[i][j] == a:
                    x1, y1 = i, j
                if matrixB[i][j] == a:
                    x2, y2 = i, j
        for i in range(3):
            for j in range(3):
                if winMatrix[i][j] == a:
                    xg, yg = i, j
        hn1 = hn1 + abs(x1 - xg) + abs(y1 - yg)
        hn2 = hn2 + abs(x2 - xg) + abs(y2 - yg)
    hn = hn1 + hn2
    return hn

cost = costCal([matrix1, matrix2])
costList = []
fatherList = []
actionList = []
stepList = []
matrixInFrontier = []
frontier = [[[matrix1, matrix2], father, action, step, cost]]
matrixInFrontier.append(frontier[0][0])
reached = []
done = False

while(not done and frontier):
    minCostMatrix = 0
    minCost = frontier[0][4]

    for i in range(1, len(frontier)):
        currCost = frontier[i][4]
        if currCost < minCost:
            minCost = currCost
            minCostMatrix = i
    node, father, action, step, cost = frontier.pop(minCostMatrix)

    matrixInFrontier.remove(node)
    reached.append(node)
    fatherList.append(father)
    actionList.append(action)
    stepList.append(step)
    costList.append(cost)

    if node[0] == winMatrix and node[1] == winMatrix:
        print("Tìm được kết quả")

        pathMatrix = []
        pathAction = []
        pathStep = []

        pathNode = node

        while pathNode is not None:
            i = reached.index(pathNode)
            pathMatrix.append(reached[i])
            pathAction.append(actionList[i])
            pathStep.append(stepList[i])
            pathNode = fatherList[i]

        pathMatrix.reverse()
        pathAction.reverse()
        pathStep.reverse()

        print("Đường đi: \n")
        for i in range(len(pathMatrix)):
            print(f"Step: {pathStep[i]}")
            print(f"Action: {pathAction[i]}")
            print("Ma trận 1:")
            for row in pathMatrix[i][0]:
                print(row)
            print("Ma trận 2:")
            for row in pathMatrix[i][1]:
                print(row)
            print("\n")

        done = True
        break

    matrixA, matrixB = node[0], node[1]

    xA, yA = -1, -1
    for i in range(3):
        for j in range(3):
            if matrixA[i][j] == 0:
                xA, yA = i, j

    xB, yB = -1, -1
    for i in range(3):
        for j in range(3):
            if matrixB[i][j] == 0:
                xB, yB = i, j

    if (xA > 0 and matrixA != winMatrix) or (xB > 0 and matrixB != winMatrix):
        newMatrixA = copy.deepcopy(matrixA)
        newMatrixB = copy.deepcopy(matrixB)

        if xA > 0 and matrixA != winMatrix:
            newMatrixA[xA][yA] = newMatrixA[xA-1][yA]
            newMatrixA[xA-1][yA] = 0

        if xB > 0 and matrixB != winMatrix:
            newMatrixB[xB][yB] = newMatrixB[xB-1][yB]
            newMatrixB[xB-1][yB] = 0

        newPair = [newMatrixA, newMatrixB]
        if newPair not in matrixInFrontier and newPair not in reached:
            newMatrixCost = costCal(newPair)
            frontier.append([newPair, node, "UP", step + 1, newMatrixCost])
            matrixInFrontier.append(newPair)

    if (xA < 2 and matrixA != winMatrix) or (xB < 2 and matrixB != winMatrix):
        newMatrixA = copy.deepcopy(matrixA)
        newMatrixB = copy.deepcopy(matrixB)

        if xA < 2 and matrixA != winMatrix:
            newMatrixA[xA][yA] = newMatrixA[xA+1][yA]
            newMatrixA[xA+1][yA] = 0

        if xB < 2 and matrixB != winMatrix:
            newMatrixB[xB][yB] = newMatrixB[xB+1][yB]
            newMatrixB[xB+1][yB] = 0

        newPair = [newMatrixA, newMatrixB]
        if newPair not in matrixInFrontier and newPair not in reached:
            newMatrixCost = costCal(newPair)
            frontier.append([newPair, node, "DOWN", step + 1, newMatrixCost])
            matrixInFrontier.append(newPair)

    if (yA > 0 and matrixA != winMatrix) or (yB > 0 and matrixB != winMatrix):
        newMatrixA = copy.deepcopy(matrixA)
        newMatrixB = copy.deepcopy(matrixB)

        if yA > 0 and matrixA != winMatrix:
            newMatrixA[xA][yA] = newMatrixA[xA][yA-1]
            newMatrixA[xA][yA-1] = 0

        if yB > 0 and matrixB != winMatrix:
            newMatrixB[xB][yB] = newMatrixB[xB][yB-1]
            newMatrixB[xB][yB-1] = 0

        newPair = [newMatrixA, newMatrixB]
        if newPair not in matrixInFrontier and newPair not in reached:
            newMatrixCost = costCal(newPair)
            frontier.append([newPair, node, "LEFT", step + 1, newMatrixCost])
            matrixInFrontier.append(newPair)

    if (yA < 2 and matrixA != winMatrix) or (yB < 2 and matrixB != winMatrix):
        newMatrixA = copy.deepcopy(matrixA)
        newMatrixB = copy.deepcopy(matrixB)

        if yA < 2 and matrixA != winMatrix:
            newMatrixA[xA][yA] = newMatrixA[xA][yA+1]
            newMatrixA[xA][yA+1] = 0

        if yB < 2 and matrixB != winMatrix:
            newMatrixB[xB][yB] = newMatrixB[xB][yB+1]
            newMatrixB[xB][yB+1] = 0

        newPair = [newMatrixA, newMatrixB]
        if newPair not in matrixInFrontier and newPair not in reached:
            newMatrixCost = costCal(newPair)
            frontier.append([newPair, node, "RIGHT", step + 1, newMatrixCost])
            matrixInFrontier.append(newPair)
```

```python
import random

def calculate_conflicts(state, var, val):
    conflicts = 0
    for i in range(9):
        if i != var and state[i] == val:
            conflicts += 1
    if var < 7:
        if val >= state[var+1]:
            conflicts += 1
    if 0 < var <= 7:
        if state[var-1] >= val:
            conflicts += 1
    if var == 8:
        for i in range(8):
            if val >= state[i]:
                conflicts += 1
    elif var < 8:
        if state[8] >= val:
            conflicts += 1
    return conflicts

def min_conflicts(max_steps=1000):
    state = [random.randint(0, 8) for _ in range(9)]
    trace = []
    trace.append(f"Trạng thái khởi tạo ban đầu: {list(state)}")

    for step in range(max_steps):
        conflicted_vars = []
        for i in range(9):
            if calculate_conflicts(state, i, state[i]) > 0:
                conflicted_vars.append(i)

        if not conflicted_vars:
            return state, step, trace

        var = random.choice(conflicted_vars)

        min_c = float('inf')
        best_vals = []

        domain_conflicts = []

        for v in range(9):
            c = calculate_conflicts(state, var, v)
            domain_conflicts.append(f"Giá trị {v}: {c} lỗi") # Ghi nhận số lỗi

            if c < min_c:
                min_c = c
                best_vals = [v]
            elif c == min_c:
                best_vals.append(v)

        chosen_val = random.choice(best_vals)
        state[var] = chosen_val

        trace.append(f"\n--- BƯỚC {step + 1} ---")
        trace.append(f"- Các ô đang vi phạm: {conflicted_vars}")
        trace.append(f"- Sửa: Ô {var}")
        trace.append(f"- Thử điền các giá trị domain vào Ô {var},xung đột:")
        trace.append(f"  | { ' | '.join(domain_conflicts) } |")
        trace.append(f"- CẬP NHẬT: Chọn điền giá trị {chosen_val} số lỗi ít nhất là {min_c})")
        trace.append(f"- Mảng sau khi cập nhật: {list(state)}")

    return None, max_steps, []

def solve_8_puzzle_min_conflicts():
    max_restarts = 500
    steps_per_try = 500

    print("Bắt đầu chạy thuật toán Min-Conflicts...\n")

    for i in range(max_restarts):
        result, steps, trace = min_conflicts(max_steps=steps_per_try)

        if result:
            print(f"LẦN RESTART THỨ {i}!")



            for line in trace:
                print(line)

            print("="*80)
            print(f"-> Số bước lặp: {steps}")
            print(f"-> Kết quả cuối cùng (Ô 0 -> 8): {result}")
            return result

    print("Không tìm thấy nghiệm trong giới hạn")
    return None

if __name__ == "__main__":
    solve_8_puzzle_min_conflicts()
```

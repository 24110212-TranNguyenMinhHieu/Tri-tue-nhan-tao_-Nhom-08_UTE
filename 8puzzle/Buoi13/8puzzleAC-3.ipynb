```python
import copy

print("Nhập domain cho 9 ô")

domains = []
for i in range(9):
    data = list(map(int, input(f"Nhập domain cho ô thứ {i}: ").split()))
    domains.append(data)

arcs = []

for i in range(7):
    arcs.append((i, i+1))
    arcs.append((i+1, i))

for i in range(8):
    arcs.append((8, i))
    arcs.append((i, 8))

arcs = list(set(arcs))
queue = copy.deepcopy(arcs)

def satisfies_rule(xi, val_i, xj, val_j):
    """Kiểm tra xem giá trị val_i của ô xi và val_j của ô xj có hợp lệ không"""
    if xi < 7 and xj == xi + 1:
        return val_i < val_j
    if xj < 7 and xi == xj + 1:
        return val_i > val_j

    if xi == 8 and xj < 8:
        return val_i < val_j
    if xj == 8 and xi < 8:
        return val_i > val_j

    return True

def revise(xi, xj):
    revised = False
    for x in domains[xi][:]:
        satisfies = False
        for y in domains[xj]:
            if satisfies_rule(xi, x, xj, y):
                satisfies = True
                break

        if not satisfies:
            domains[xi].remove(x)
            revised = True

    return revised

done = False
success = True
step = 0
loop_count = 0

while queue and not done:
    loop_count += 1
    xi, xj = queue.pop(0)

    print(f"\n--- Lần lặp {loop_count}: ---")
    print(f"Lấy cung ({xi}, {xj}) ra khỏi queue.")

    domain_before = list(domains[xi])

    if revise(xi, xj):
        step += 1
        print(f"-> Domain của Ô {xi} thay đổi: {domain_before} ---> {domains[xi]}")

        if len(domains[xi]) == 0:
            print(f" Ô {xi} rỗng, bài toán vô nghiệm.")
            success = False
            done = True
            break

        added_arcs = []
        for xk in range(9):
            if xk != xi and xk != xj:
                if (xk, xi) in arcs and (xk, xi) not in queue:
                    queue.append((xk, xi))
                    added_arcs.append((xk, xi))

        if added_arcs:
            print(f"-> [LAN TRUYỀN] Đã thêm các cung sau vào lại queue: {added_arcs}")
    else:
        print(f"-> [BỎ QUA] Không có cắt tỉa nào. Domain Ô {xi} giữ nguyên là {domains[xi]}")

if not success:
    print("Bài toán vô nghiệm (có ô không thể chọn được số nào hợp lệ).")
else:
    print(f"Thuật toán AC-3 hoàn tất sau {step} lần cắt tỉa thành công (tổng số lần xét queue: {loop_count}).")
    print("\nTập giá trị (Domain) của các ô sau khi thu gọn:")
    for i in range(9):
        print(f"Ô {i}: {domains[i]}")

    is_solved = True
    for d in domains:
        if len(d) != 1:
            is_solved = False
            break

    if is_solved:
        print("\nAC-3 đã ép được.")
        print("Kết quả (Ô 0 -> 8) là:")
        print([domains[i][0] for i in range(9)])
```

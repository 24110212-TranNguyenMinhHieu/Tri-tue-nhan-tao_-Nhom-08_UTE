```python
def is_goal(state, goals):
    for goal in goals:
        if all(matrix == goal for matrix in state):
            return True
    return False
```

```python
import copy

def get_neighbors(node, goals):

    actions = [("UP", -1, 0), ("DOWN", 1, 0), ("LEFT", 0, -1), ("RIGHT", 0, 1)]
    neighbors = []

    for act_name, dx, dy in actions:
        new_node = []
        moved_any = False

        for matrix in node:
            if matrix in goals:
                new_node.append(copy.deepcopy(matrix))
                continue

            x = y = -1
            for i in range(3):
                for j in range(3):
                    if matrix[i][j] == 0: x, y = i, j

            nx, ny = x + dx, y + dy
            if 0 <= nx < 3 and 0 <= ny < 3:
                new_matrix = copy.deepcopy(matrix)
                new_matrix[x][y] = new_matrix[nx][ny]
                new_matrix[nx][ny] = 0
                new_node.append(new_matrix)
                moved_any = True
            else:
                new_node.append(copy.deepcopy(matrix))

        if moved_any:
            neighbors.append([new_node, act_name])

    return neighbors
```

```python
def print_solution(result, step_count=1):
    if not result or result == "failure":
        return

    action = result[0]
    plan = result[1]

    print(f"\nBước {step_count}: Di chuyển '{action}'")

    if not plan:
        return

    for state_and_plan in plan:
        state = state_and_plan[0]
        next_plan = state_and_plan[1]

        matrix = state[0]
        for row in matrix:
            print(row)

        if next_plan == []:
            print(" Đã tới đích!")
        else:
            print_solution(next_plan, step_count + 1)
```

```python
import random

OPPOSITE_ACTIONS = {
    "UP": "DOWN",
    "DOWN": "UP",
    "LEFT": "RIGHT",
    "RIGHT": "LEFT"
}

def backtracking_feedforward(state, goals, path, last_action=None, depth_limit=35):

    if is_goal(state, goals):
        return []

    if state in path or len(path) >= depth_limit:
        return "failure"

    neighbors = get_neighbors(state, goals)
    random.shuffle(neighbors)

    for new_state, action in neighbors:

        if last_action is not None and action == OPPOSITE_ACTIONS.get(last_action):
            continue

        if new_state not in path:
            plan = backtracking_feedforward(new_state, goals, path + [state], action, depth_limit)

            if plan != "failure":
                return [action, [[new_state, plan]]]

    return "failure"
```

```python
winMatrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0],
]

print("Ma trận đích:")
for row in winMatrix:
    print(row)

data_input = input("Nhập ma trận (ví dụ: 1 2 3 4 0 6 7 5 8): ")
if not data_input:
    data = [1, 2, 3, 4, 0, 6, 7, 5, 8]
else:
    data = list(map(int, data_input.split()))

matrix = [data[i:i+3] for i in range(0, 9, 3)]

print("\nMa trận bắt đầu:")
for row in matrix:
    print(row)

start_state = [matrix]
goals = [winMatrix]

result = backtracking_feedforward(start_state, goals, path=[], depth_limit=35)

if result == "failure":
    print("\nKhông tìm thấy đường đi tới đích trong giới hạn độ sâu cho phép!")
else:
    print("\nĐÃ TÌM THẤY GIẢI PHÁP:")
    print_solution(result)
```

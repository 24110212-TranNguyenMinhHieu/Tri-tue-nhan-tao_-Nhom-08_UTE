```python
def is_goal(state, goals):
    for goal in goals:
        if all(matrix == goal for matrix in state):
            return True
    return False
```

```python
import copy

def get_neighbors(node, goals):

    actions = [("UP", -1, 0), ("DOWN", 1, 0), ("LEFT", 0, -1), ("RIGHT", 0, 1)]
    neighbors = []

    for act_name, dx, dy in actions:
        new_node = []
        moved_any = False

        for matrix in node:
            if matrix in goals:
                new_node.append(copy.deepcopy(matrix))
                continue

            x = y = -1
            for i in range(3):
                for j in range(3):
                    if matrix[i][j] == 0: x, y = i, j

            nx, ny = x + dx, y + dy
            if 0 <= nx < 3 and 0 <= ny < 3:
                new_matrix = copy.deepcopy(matrix)
                new_matrix[x][y] = new_matrix[nx][ny]
                new_matrix[nx][ny] = 0
                new_node.append(new_matrix)
                moved_any = True
            else:
                new_node.append(copy.deepcopy(matrix))

        if moved_any:
            neighbors.append([new_node, act_name])

    return neighbors
```

```python
import copy
import random

def and_or_graph_search(start_state, goals):
    return or_search(start_state, goals, [])

def or_search(state, goals, path):
    if is_goal(state, goals):
        return []

    if state in path:
        return "failure"

    neighbors = get_neighbors(state, goals)

    random.shuffle(neighbors)

    for new_state, action in neighbors:

        result_states = [new_state]

        plan = and_search(result_states, goals, path + [state])

        if plan != "failure":
            return [action, plan]

    return "failure"

def and_search(states, goals, path):
    plans = []

    for state in states:
        plan_s = or_search(state, goals, path)

        if plan_s == "failure":
            return "failure"
        plans.append([state, plan_s])

    return plans
```

```python
def print_solution(result, step_count=1):
    if not result or result == "failure":
        return

    action = result[0]
    plan = result[1]

    print(f"\nBước {step_count}: Di chuyển '{action}'")

    if not plan:
        return

    for state_and_plan in plan:
        state = state_and_plan[0]
        next_plan = state_and_plan[1]

        matrix = state[0]
        for row in matrix:
            print(row)

        if next_plan == []:
            print(" Đã tới đích!")
        else:
            print_solution(next_plan, step_count + 1)
```

```python
winMatrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0],
]

print("Ma trận đích:")
for row in winMatrix:
    print(row)

data = list(map(int, input("Nhập ma trận: ").split()))
matrix = [data[i:i+3] for i in range(0, 9, 3)]

print("\nMa trận bắt đầu:")
for row in matrix:
    print(row)

start_state = [matrix]
goals = [winMatrix]

result = and_or_graph_search(start_state, goals)

if result == "failure":
    print("Không tìm thấy đường đi tới đích!")
else:
    print("ĐÃ TÌM THẤY GIẢI PHÁP:")
    print_solution(result)
```

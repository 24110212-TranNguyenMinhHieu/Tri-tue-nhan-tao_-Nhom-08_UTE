```python
import random

winMatrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0],
]

gameACTION = ["UP", "DOWN", "LEFT", "RIGHT"]

data = list(map(int, input("Nhập ma trận: ").split()))
matrix = [data[i:i+3] for i in range(0, 9, 3)]

def findZero(matrix):
    for i in range(3):
        for j in range(3):
            if matrix[i][j] == 0:
                x = i
                y = j
    return x, y

def getRules(x, y, gameACTION):
    ruleAction = []
    if x > 0:
        ruleAction.append(gameACTION[0])
    if x < 2:
        ruleAction.append(gameACTION[1])
    if y > 0:
        ruleAction.append(gameACTION[2])
    if y < 2:
        ruleAction.append(gameACTION[3])
    return ruleAction

def move(matrix, x, y, zeroAction):
    if zeroAction == "UP":
        matrix[x][y] = matrix[x-1][y]
        matrix[x-1][y] = 0
    if zeroAction == "DOWN":
        matrix[x][y] = matrix[x+1][y]
        matrix[x+1][y] = 0
    if zeroAction == "LEFT":
        matrix[x][y] = matrix[x][y-1]
        matrix[x][y-1] = 0
    if zeroAction == "RIGHT":
        matrix[x][y] = matrix[x][y+1]
        matrix[x][y+1] = 0
    return matrix

count = 0
while (matrix != winMatrix and count < 100):
    x, y = findZero(matrix)
    ruleAction = getRules(x, y, gameACTION)
    zeroAction = random.choice(ruleAction)
    matrix = move(matrix, x, y, zeroAction)

    count += 1
    print("Bước", count, ":", zeroAction)
    for row in matrix:
        print(row)

if (matrix == winMatrix):
    print("Đã hoàn thành")
else:
    print("Dừng lại sau 100 bước")
```

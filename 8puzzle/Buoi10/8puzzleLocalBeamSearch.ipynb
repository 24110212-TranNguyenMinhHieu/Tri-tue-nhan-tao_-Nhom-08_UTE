```python
import random
import copy
import sys

winMatrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0],
]

data = list(map(int, input("Nhập ma trận: ").split()))
matrix = [data[i:i+3] for i in range(0, 9, 3)]
k = int(input("Nhập k: "))

startMatrix = matrix
father = None
action = None
step = 0

def costCal(thisMatrix):
    hn = 0
    x1 = x2 = y1 = y2 = 0
    for a in range(1,9):
        for i in range(3):
            for j in range(3):
                if thisMatrix[i][j] == a:
                    x1, y1 = i, j
        for i in range(3):
            for j in range(3):
                if winMatrix[i][j] == a:
                    x2, y2 = i, j
        hn = hn + abs(x1 - x2) + abs(y1 - y2)
    return hn

cost = costCal(matrix)

info = [[matrix, father, action, step, cost]]
currentStateSet = [info]

visited = [matrix]

done = False
while(not done):
    neighborStates = []

    for currState in currentStateSet:
        lastStepInfo = currState[-1]
        currMatrix = lastStepInfo[0]
        currStep = lastStepInfo[3]

        for i in range(3):
            for j in range(3):
                if currMatrix[i][j] == 0:
                    x, y = i, j

        if x > 0:
            newMatrix = copy.deepcopy(currMatrix)
            newMatrix[x][y] = newMatrix[x-1][y]
            newMatrix[x-1][y] = 0
            if newMatrix not in visited:
                visited.append(newMatrix)
                childCost = costCal(newMatrix)
                newState = copy.deepcopy(currState)
                newState.append([newMatrix, currMatrix, "UP", currStep + 1, childCost])
                neighborStates.append(newState)

        if x < 2:
            newMatrix = copy.deepcopy(currMatrix)
            newMatrix[x][y] = newMatrix[x+1][y]
            newMatrix[x+1][y] = 0
            if newMatrix not in visited:
                visited.append(newMatrix)
                childCost = costCal(newMatrix)
                newState = copy.deepcopy(currState)
                newState.append([newMatrix, currMatrix, "DOWN", currStep + 1, childCost])
                neighborStates.append(newState)

        if y > 0:
            newMatrix = copy.deepcopy(currMatrix)
            newMatrix[x][y] = newMatrix[x][y-1]
            newMatrix[x][y-1] = 0
            if newMatrix not in visited:
                visited.append(newMatrix)
                childCost = costCal(newMatrix)
                newState = copy.deepcopy(currState)
                newState.append([newMatrix, currMatrix, "LEFT", currStep + 1, childCost])
                neighborStates.append(newState)

        if y < 2:
            newMatrix = copy.deepcopy(currMatrix)
            newMatrix[x][y] = newMatrix[x][y+1]
            newMatrix[x][y+1] = 0
            if newMatrix not in visited:
                visited.append(newMatrix)
                childCost = costCal(newMatrix)
                newState = copy.deepcopy(currState)
                newState.append([newMatrix, currMatrix, "RIGHT", currStep + 1, childCost])
                neighborStates.append(newState)

    if len(neighborStates) == 0:
        m = len(currentStateSet)
        for i in range(m - 1):
            for j in range(i + 1, m):
                costI = currentStateSet[i][-1][4]
                costJ = currentStateSet[j][-1][4]
                if costI > costJ:
                    temp = currentStateSet[i]
                    currentStateSet[i] = currentStateSet[j]
                    currentStateSet[j] = temp

        bestInfo = currentStateSet[0]

        print("Không đi được nữa")
        print("Start:")
        for row in bestInfo[0][0]:
            print(row)
        print("\n")
        print("Goal")
        for row in winMatrix:
            print(row)
        print("\n")
        print("CÁC BƯỚC DI CHUYỂN TỐT NHẤT")
        for i in range(0, len(bestInfo)):
            stepInfo = bestInfo[i]
            currMatrixStep = stepInfo[0]
            currAct = stepInfo[2]
            currStepCount = stepInfo[3]
            currCost = stepInfo[4]

            print(f"Step {currStepCount}:   Action = {currAct}   Cost: {currCost}")
            for row in currMatrixStep:
                print(row)
        sys.exit()

    for neighbor in neighborStates:
        if neighbor[-1][0] == winMatrix:
            print("KẾT QUẢ ")
            print("Start:")
            for row in neighbor[0][0]:
                print(row)
            print("\n")
            print("Goal")
            for row in winMatrix:
                print(row)
            print("\n")
            print("CÁC BƯỚC DI CHUYỂN")
            for i in range(0, len(neighbor)):
                stepInfo = neighbor[i]
                currMatrixStep = stepInfo[0]
                currAct = stepInfo[2]
                currStepCount = stepInfo[3]
                currCost = stepInfo[4]

                print(f"Step {currStepCount}:   Action = {currAct}   Cost: {currCost}")
                for row in currMatrixStep:
                    print(row)
            sys.exit()

    n = len(neighborStates)
    for i in range(n - 1):
        for j in range(i + 1, n):
            costI = neighborStates[i][-1][4]
            costJ = neighborStates[j][-1][4]
            if costI > costJ:
                temp = neighborStates[i]
                neighborStates[i] = neighborStates[j]
                neighborStates[j] = temp

    currentStateSet = []

    for i in range(k):
        if len(neighborStates) > 0:
            currentStateSet.append(neighborStates.pop(0))
```

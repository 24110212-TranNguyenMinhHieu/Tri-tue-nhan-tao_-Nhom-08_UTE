```python
import random
import copy
import sys

winMatrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0],
]

data = list(map(int, input("Nhập ma trận: ").split()))
matrix = [data[i:i+3] for i in range(0, 9, 3)]

startMatrix = matrix
father = None
action = None
step = 0

def costCal(thisMatrix):
    hn = 0
    x1 = x2 = y1 = y2 = 0
    for a in range(1,9):
        for i in range(3):
            for j in range(3):
                if thisMatrix[i][j] == a:
                    x1, y1 = i, j
        for i in range(3):
            for j in range(3):
                if winMatrix[i][j] == a:
                    x2, y2 = i, j
        hn = hn + abs(x1 - x2) + abs(y1 - y2)
    return hn

cost = costCal(matrix)

info = [[matrix, father, action, step, cost]]
betterNeighbor = []

maxRestart = 50

done = False
for iRestart in range(1, maxRestart + 1):
    print(f"\n Lượt thứ {iRestart}")

    matrix = copy.deepcopy(startMatrix)
    cost = costCal(startMatrix)
    action = None
    step = 0
    info = [[matrix, None, action, step, cost]]

    done = False
    while(not done):
        if info[-1][0] == winMatrix:
            done = True
            print("KẾT QUẢ ")
            print("Start:")
            for row in info[0][0]:
                print(row)

            print("\n")

            print("Goal")
            for row in winMatrix:
                print(row)

            print("\n")

            print("CÁC BƯỚC DI CHUYỂN")
            for i in range(0, len(info)):
                currentState = info[i]
                currMatrix = currentState[0]
                currAct = currentState[2]
                currStep = currentState[3]
                currCost = currentState[4]

                print(f"Step {currStep}:   Action = {currAct}   Cost: {currCost}")
                for row in currMatrix:
                    print(row)
            sys.exit()

        for i in range(3):
            for j in range(3):
                if matrix[i][j] == 0:
                    x, y = i, j

        moved = False

        if x > 0:
            newMatrix = copy.deepcopy(matrix)
            newMatrix[x][y] = newMatrix[x-1][y]
            newMatrix[x-1][y] = 0
            childCost = costCal(newMatrix)
            if childCost < cost:
                betterNeighbor.append([newMatrix, matrix, "UP", step + 1, childCost])
                moved = True

        if x < 2:
            newMatrix = copy.deepcopy(matrix)
            newMatrix[x][y] = newMatrix[x+1][y]
            newMatrix[x+1][y] = 0
            childCost = costCal(newMatrix)
            if childCost < cost:
                betterNeighbor.append([newMatrix, matrix, "DOWN", step + 1, childCost])
                moved = True

        if y > 0:
            newMatrix = copy.deepcopy(matrix)
            newMatrix[x][y] = newMatrix[x][y-1]
            newMatrix[x][y-1] = 0
            childCost = costCal(newMatrix)
            if childCost < cost:
                betterNeighbor.append([newMatrix, matrix, "LEFT", step + 1, childCost])
                moved = True

        if y < 2:
            newMatrix = copy.deepcopy(matrix)
            newMatrix[x][y] = newMatrix[x][y+1]
            newMatrix[x][y+1] = 0
            childCost = costCal(newMatrix)
            if childCost < cost:
                betterNeighbor.append([newMatrix, matrix, "RIGHT", step + 1, childCost])
                moved = True
        if betterNeighbor:
            nextState = random.choice(betterNeighbor)
            matrix = nextState[0]
            step = nextState[3]
            cost = nextState[4]
            info.append(nextState)
            betterNeighbor = []

        if moved == False:
            done = True
            if iRestart == maxRestart:
                print("KẾT QUẢ ")
                print("Start:")
                for row in info[0][0]:
                    print(row)

                print("\n")

                print("Goal")
                for row in winMatrix:
                    print(row)

                print("\n")

                print("CÁC BƯỚC DI CHUYỂN")
                for i in range(0, len(info)):
                    currentState = info[i]
                    currMatrix = currentState[0]
                    currAct = currentState[2]
                    currStep = currentState[3]
                    currCost = currentState[4]

                    print(f"Step {currStep}:   Action = {currAct}   Cost: {currCost}")
                    for row in currMatrix:
                        print(row)
                    print("\n")
                print(f"\n{maxRestart} lượt mà không chạm được Goal.")
```
